# Linear regression
Given the dataset x, y apply linear regression to find the model of the form: y = ax + b

What are the values of "a" and "b" in that model?

Recall: 
```{Python}
x_reshaped = x.reshape(-1, 1) #fit needs data in this shape
```

````python
X = np.loadtxt('lr_x.txt', dtype=float)
y = np.loadtxt('lr_y.txt', dtype=float)

# X must be 2D: (n_samples, 1)
X = X.reshape(-1, 1)        

reg = LinearRegression().fit(X, y)

a = reg.coef_[0]      # slope
b = reg.intercept_    # intercept

print(a, b)
````

````python

x = np.loadtxt('lr_x.txt', dtype=float)
y = np.loadtxt('lr_y.txt', dtype=float)

# Compute line values
y_pred = a * x + b

plt.figure(figsize=(8, 6))
plt.plot(x, y, 'ro', label='Data')         # scatter
plt.plot(x, y_pred, 'b-', label='Fit')     # regression line
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.show()
````
