### Intermediate step 1:

Implement a function that takes two equal size images and calculates the sum of absolute differences. The two equal-sized images are nose_left.png (used as the template) and nose_right.png.

```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
from skimage import data
import numpy as np
```

```python
nose_left = cv2.imread('nose_left.png')
nose_right = cv2.imread('nose_right.png')

nose_left_rgb = cv2.cvtColor(nose_left, cv2.COLOR_BGR2RGB)
nose_left_gray = cv2.cvtColor(nose_left_rgb, cv2.COLOR_RGB2GRAY)

nose_right_rgb = cv2.cvtColor(nose_right, cv2.COLOR_BGR2RGB)
nose_right_gray = cv2.cvtColor(nose_right_rgb, cv2.COLOR_RGB2GRAY)

fig, axes = plt.subplots(ncols=2, figsize=(10,10))

print(nose_left_gray.shape)

axes[0].imshow(nose_left_gray, 'gray')
axes[0].set_title("left nose")
axes[0].axis("off") 

axes[1].imshow(nose_right_gray, 'gray')
axes[1].set_title("right nose")
axes[1].axis("off")
```

```python
def sum_abs_diff(img1, img2):
    return np.sum(abs(img1-img2))

print(sum_abs_diff(nose_left_gray, nose_right_gray))
```
